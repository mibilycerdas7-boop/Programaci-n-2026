## 1. ¿Qué hacen `while True` y `break`?

Un ciclo `while` repite instrucciones mientras una condición sea verdadera.
Cuando escribimos `while True`, la condición siempre es verdadera, así que el
ciclo necesita una salida explícita. Esa salida se logra con `break`.

Use este patrón cuando no sabe de antemano cuántas veces se repetirá una
acción. Por ejemplo: pedir una opción válida, esperar una palabra correcta o
continuar un juego hasta que el usuario acierte.


## 2. Ejemplo corto: repetir hasta escribir una palabra

Ejecute el ejemplo y observe dónde aparece la salida del ciclo.


````python
while True:
    palabra = input("Escriba salir para terminar: ").strip().lower()

    if palabra == "salir":
        print("Programa finalizado.")
        break

    print("Todavía no termina.")
````

## 3. Ejemplo corto: repetir hasta cumplir una condición numérica

Este ejemplo usa `int()` porque `input()` siempre recibe texto.


````python
while True:
    numero = int(input("Escriba un número mayor que 10: "))

    if numero > 10:
        print("Número aceptado.")
        break

    print("Debe intentar otra vez.")
````

## 6. Consigna: Adivina el número

Cree un juego llamado **Adivina el número: frío o caliente**.

El programa debe:

1. Generar un número secreto al azar entre 1 y 20 usando `random.randint()`.
2. Pedir intentos al usuario hasta que acierte.
3. Convertir cada intento a número entero.
4. Comparar el intento con el número secreto.
5. Dar pistas de dirección: si el número secreto es mayor o menor que el intento.
6. Dar pistas de temperatura según la distancia al número secreto.
7. Felicitar al usuario cuando acierte.
8. Mostrar cuántos intentos realizó el usuario.
9. Terminar el ciclo con `break` cuando el usuario acierte.

### Reglas de temperatura

Use la distancia entre el intento y el número secreto para crear pistas más
creativas. La distancia debe ser siempre positiva.

Para calcular una distancia positiva puede usar `abs()`. Por ejemplo,
`abs(3 - 8)` produce `5`, no `-5`.

Puede usar esta escala o proponer una equivalente:

| Distancia al número secreto | Pista sugerida |
|---:|---|
| `0` | Acertó. Termina el juego. |
| `1` o `2` | Hirviendo: está casi encima del número. |
| `3` a `5` | Caliente: está cerca. |
| `6` a `10` | Tibio: va por buen camino, pero falta. |
| Más de `10` | Frío: está lejos. |

Ejemplo de experiencia esperada:

## 7. Código del estudiante

Escriba aquí su solución completa.


````python
# Escriba aquí su juego Adivina el número.
import random

#Genere el número secreto al azar entre 1 al 20 una sola vez al inicio
numero_secreto = random.randint(1, 20)

#Inicie el conatador de eventos
intentos = 0

print("¡Bienvenidos al juego: Alimenta al Mostrito Raynow!")
print("Mi mostruito tiene hambre y piensa en una cantidad del 1 al 20. ¡Adivina cuánto quiere comer!\n")

while True:
    #solicitemos el intento y lo convierte a entero.
    intento = int(input("cantidad de comida ofrecida: "))

    # Aumentamos el contador de intentos (corregido sin espacio: +=)
    intentos += 1

    #Vweificamos si acertó
    if intento == numero_secreto:
        print(f"¡Delicioso! le diste justo en el blanco. Adivinaste en {intentos} intentos.")
        break  # Rompe el ciclo para finalizar el juego

    #Damos pistas de dirreción (mayor o menor)
    elif intento < numero_secreto:
        print("El mostrito sigue con hambre... ¡necesita más comida!\n")
    else:
        print("¡Te pasaste! El mostrito está muy lleno, ofrece menos comida.\n")

    # Calculemos la distancia positiva usando abs() (corregido 'distacia' a 'distancia')
    distancia = abs(intento - numero_secreto)
    if distancia <= 2 :
        print("Pista: ¡Ñam! Estoy a punto de comer exactamente lo que deseaba.\n")
    elif distancia <= 5 :
        print("Pista: Su estómago empieza a rugir más cerca del número correcto.\n")
    elif distancia <= 10 :
        print("Pista: Todavía me falta bastante comida para quedar satisfecho.\n")
    else:
        print("Pista: Estoy completamente desorientado, la cantidad está muy lejos.\n")
````

1. ¿Por qué este programa necesita repetir instrucciones?

El juego requiere repetir instrucciones (usando el ciclo while True) por una razón fundamental: no sabemos cuántos intentos le tomará al usuario adivinar el número secreto.
Dado que el juego consiste en adivinar una cantidad del 1 al 20, ocurren las siguientes situaciones que exigen repetición:

A.	La incertidumbre de la respuesta: Como es muy probable que el usuario no acierte en su primer intento, el programa necesita volver a preguntar y evaluar la respuesta tantas veces como sea necesario.

B.	Dar pistas progresivas: Cada vez que el usuario ingresa un número erróneo, el programa necesita calcular la distancia y mostrar una pista nueva ("está muy lejos", "más comida", etc.). Sin una estructura repetitiva, el programa terminaría de inmediato después de la primera entrada sin darle oportunidad al usuario de corregir.

C.	Controlar el fin del juego: El ciclo se mantiene activo de manera indefinida hasta que el usuario acierta exactamente el número, momento en el cual la instrucción break rompe la repetición y finaliza el juego con éxito.
