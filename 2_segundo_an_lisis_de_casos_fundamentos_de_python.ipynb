# Problema #1: Calculadora de propina inteligente

## Problema

Desarrollar un programa que calcule de forma automática el valore de la propina y el total a pagar de una cuenta, basada en el monto ingresado y el porcentaje deseado por el usuario.

## Entrada

- El monto total de la cuenta y el porcentaje de propina.

## Proceso

- validar datos ingresados, sean númericos mediante un bloque  try_except. El calcular la propina multiplicando el monto por el porcentaje, dividido entre 100 y sumar el resultado al total inicial para obtener el monto final.

## Salida esperada

- Mantener el desglose claro de la propina calculada y el total final a pagar utilizando f-strimgs

```python
# Escriba la solución aquí #Calculadora de propina inteligente
try:
    cuenta = float(input("Ingrese el monto total de la cuenta: "))
    porcentaje = float(input("Ingrese el porcentaje de propina deseado (ej. 10, 15): "))

    propina = cuenta * (porcentaje / 100)
    total_final = cuenta + propina

    print(f"\n--- Resumen de la Cuenta ---")
    print(f"Monto base: ${cuenta:.2f}")
    print(f"Propina ({porcentaje}%): ${propina:.2f}")
    print(f"Total a pagar: ${total_final:.2f}")
except ValueError:
    print("Error: Por favor, ingrese valores numéricos válidos.")
```

# Problema #2: Detector de número par/impar

## Problema

Crear un progama que determine si un número entero ingresado por el usuario es par o impar.

## Entrada

- Número entero

## Proceso

- Utilizando bloque try-except para asegurar que los datos sean números validos enteros  (int). emplear el operador modulo (%2) para comprobar si el residuo de dividir entre dos es igual a cero

## Salida esperada

- un mensaje en pantalla indicando si el numero es par o impar.

```python
# Escriba la solución aquí  #Detector de número par/impar

# Detector de número par/impar
try:
    numero = int(input("Ingrese un número entero: "))

    if numero % 2 == 0:
        print(f"El número {numero} es PAR.")
    else:
        print(f"El número {numero} es IMPAR.")
except ValueError:
    print("Error: Debe ingresar obligatoriamente un número entero.")
```

# Problema #3: Clasificador de edad

## Problema

Diseñar un sistema que clasifique a las personas segun su rango de edad, segun el valor ingresado.

## Entrada

- la edad de las personas en años.

## Proceso

- Entradas númericas con try-except . evaluar mediante una estructura condicional (if, elif, else) los rangos correspondientes (menor edad, adulto y adulto mayor).

## Salida esperada

- Un mesjae que indique la clasificación obtenida, de la edad exacta.

```python
# Escriba la solución aquí #Clasificador de edad
try:
    edad = int(input("Ingrese su edad: "))

    if edad < 0:
        print("Error: La edad no puede ser negativa.")
    elif edad < 18:
        print(f"Con {edad} años, usted es clasificado como: Menor de edad.")
    elif edad < 65:
        print(f"Con {edad} años, usted es clasificado como: Adulto.")
    else:
        print(f"Con {edad} años, usted es clasificado como: Adulto mayor.")
except ValueError:
    print("Error: Ingrese un valor numérico entero válido para la edad.")
```

# Problema #4: Validador de compra con descuento

## Problema

Calcular el valor final de una compra compra aplicando reglas de descuentos basadas en el monto total y el comprar es cliente frecuente.

## Entrada

- Nombre del clientez, monto dela compra y condición cliente frecuente (si/no)

## Proceso

- Limpiar y formatear las cadenas de texto con .strip() y .lower() .validar el monto con try-except y aplicar operadores logicos con (and y or) para calcular descuentos especialessi el monto supera un limite o si se confirma la condición de cliente frecuente

## Salida esperada

- Un resumen personalisado con f-strings que muestre el descuento aplicado y valor total a cancelar.

## Pruebas manuales

Caso 1: Cliente frecuente, compra de $120. Descuento aplicado correctamente.

Caso 2: No frecuente, compra de $40. Sin descuento.

```python
# Escriba la solución aquí
# Validador de compra con descuento
try:
    nombre = input("Ingrese el nombre del cliente: ").strip().title()
    monto_compra = float(input("Ingrese el monto de la compra: "))
    es_frecuente = input("¿Es cliente frecuente? (s/n): ").strip().lower() == 's'

    descuento = 0.0

    if monto_compra >= 100 or es_frecuente:
        descuento = 0.10  # 10% de descuento
        print("¡Aplica para descuento especial!")
    else:
        print("No aplican descuentos para esta compra.")

    monto_final = monto_compra * (1 - descuento)

    print(f"\n--- Factura Final para {nombre} ---")
    print(f"Monto original: ${monto_compra:.2f}")
    print(f"Descuento aplicado: {int(descuento * 100)}%")
    print(f"Total a pagar: ${monto_final:.2f}")
except ValueError:
    print("Error en los datos ingresados. Asegúrese de colocar números en el monto.")
```
